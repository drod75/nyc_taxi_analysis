# Data Exploration

## Setup

```python
# importing libraries, pandas for data manipulation, requests for data gathering, colab for environment variables
import pandas as pd
import geopandas
from google.colab import userdata
from google.colab import drive
from pprint import pprint as print
```

```python
# mounting drive for future use
drive.mount('drive')
```

```python
# loading data via pandas and geo data via geo pandas
taxi_df = pd.read_csv("/content/drive/MyDrive/taxi_summer_data.csv")
zone_df = geopandas.read_file("/content/drive/MyDrive/taxi_zones_data.json")
```

```python
# mering our tables for easier use
taxi_zone_df = pd.merge(taxi_df, zone_df, left_on='pulocationid', right_on='locationid', how='left')
taxi_zone_df = taxi_zone_df.rename(columns={'borough': 'puborough', 'zone': 'puzone'})
taxi_zone_df = pd.merge(taxi_zone_df, zone_df, left_on='dolocationid', right_on='locationid', how='left')
taxi_zone_df = taxi_zone_df.rename(columns={'borough': 'duborough', 'zone': 'duzone'})
```

## Data Review

```python
# checking our data full info
taxi_zone_df.info()
```

```python
# checking the first 5 rows of our data
taxi_zone_df.head()
```

## Univariate Categorical Analysis

```python
# checking pickup borough value counts
taxi_zone_df['puborough'].value_counts()
```

```python
# checking pickup zone value counts
taxi_zone_df['puzone'].value_counts()
```

```python
# checking dropoff borough value counts
taxi_zone_df['duborough'].value_counts()
```

```python
# checking dropoff zone value counts
taxi_zone_df['duzone'].value_counts()
```

## Univariate Numerical Analysis

```python
# checking numerical values
taxi_zone_df.describe().T
```

## Categorical to Categorical

```python
# checking pickup and dropoff borough cross tab
borough_crosstab = pd.crosstab(taxi_zone_df['puborough'], taxi_zone_df['duborough'])
borough_crosstab
```

## Categorical to Numerical Analysis

```python
# checking pickup borough count 
rides_from = taxi_zone_df.groupby('puborough')['puborough'].count()
rides_from
```

```python
# cehcking dropoff borough count
rides_to = taxi_zone_df.groupby('duborough')['duborough'].count()
rides_to
```

```python
# checking ride fare/passenger info for pickup
rides_from = taxi_zone_df.groupby('puborough')[['total_amount', 'passenger_count', 'trip_distance']].describe().T
print(rides_from)
```

```python
# checking ride fare/passenger info for dropoff
rides_to = taxi_zone_df.groupby('duborough')[['total_amount', 'passenger_count', 'trip_distance']].describe().T
print(rides_to)
```
